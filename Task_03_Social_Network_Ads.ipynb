```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/Social_Network_Ads[1].csv')

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df['Purchased'].value_counts()
```

```python
plt.figure(figsize=(6,4))

df['Purchased'].value_counts().plot(kind='bar')

plt.title('Purchase Distribution')
plt.xlabel('Purchased (0 = No, 1 = Yes)')
plt.ylabel('Number of Users')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.boxplot(x='Purchased', y='Age', data=df)

plt.title('Age Distribution by Purchase')
plt.xlabel('Purchased (0 = No, 1 = Yes)')
plt.ylabel('Age')
plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.boxplot(x='Purchased', y='EstimatedSalary', data=df)

plt.title('Estimated Salary Distribution by Purchase')
plt.xlabel('Purchased (0 = No, 1 = Yes)')
plt.ylabel('Estimated Salary')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='Age',
    y='EstimatedSalary',
    hue='Purchased',
    data=df
)

plt.title('Age vs Estimated Salary by Purchase')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.show()
```

```python
plt.figure(figsize=(6,5))

corr = df[['Age', 'EstimatedSalary', 'Purchased']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap')
plt.show()
```

## Conclusion

The exploratory data analysis shows that Age and Estimated Salary are important factors related to product purchase behavior.

Older users were more likely to purchase the product, with Age showing a strong positive correlation of 0.62 with Purchased. Estimated Salary also showed a positive correlation of 0.36 with Purchased, indicating that users with higher salaries were more likely to purchase.

The analysis also showed that the dataset contains 400 records, with 143 users purchasing the product and 257 users not purchasing it.

Overall, the visualizations and correlation analysis provide useful insights into customer purchasing behavior and can help in building a machine learning classification model.
